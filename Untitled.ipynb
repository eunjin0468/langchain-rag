```python
# 라이브러리 임포트
import requests
import pandas as pd
import time
from io import StringIO
from bs4 import BeautifulSoup
from langchain_text_splitters import RecursiveCharacterTextSplitter
import yfinance as yf
import plotly.graph_objects as go
```

```python
#종목 검색 함수
def is_korean(text):
    return any("\uAC00" <= c <= "\uD7A3" for c in text)

def get_stock_code(stock_name):
    url = "https://kind.krx.co.kr/corpgeneral/corpList.do?method=download"
    response = requests.get(url, headers={"User-Agent": "Mozilla/5.0"})
    response.encoding = "euc-kr"
    df = pd.read_html(StringIO(response.text))[0]
    df["종목코드"] = df["종목코드"].astype(str).str.strip()
    result = df[df["회사명"] == stock_name]["종목코드"]
    return result.values[0] if not result.empty else None

def translate_to_english(korean_name):
    url = "https://translate.googleapis.com/translate_a/single"
    params = {"client": "gtx", "sl": "ko", "tl": "en", "dt": "t", "q": korean_name}
    try:
        response = requests.get(url, params=params)
        return response.json()[0][0][0]
    except:
        return None

def get_yahoo_ticker(query):
    url = "https://query2.finance.yahoo.com/v1/finance/search"
    params = {"q": query, "lang": "en-US", "region": "US"}
    headers = {"User-Agent": "Mozilla/5.0"}
    try:
        response = requests.get(url, params=params, headers=headers)
        quotes = response.json().get("quotes", [])
        return quotes[0]["symbol"] if quotes else None
    except:
        return None

def search_stock(stock_name):
    if stock_name.isdigit() and len(stock_name) == 6:
        print(f"✅ [국내] 종목코드 직접 입력 → {stock_name}")
        return {"type": "KR", "code": stock_name, "name": stock_name}
    
    if is_korean(stock_name):
        code = get_stock_code(stock_name)
        if code:
            print(f"✅ [국내] {stock_name} → {code}")
            return {"type": "KR", "code": code, "name": stock_name}
        
        print(f"🔄 '{stock_name}' 번역 중...")
        english_name = translate_to_english(stock_name)
        print(f"🔄 번역 결과: {english_name}")
        
        if english_name:
            ticker = get_yahoo_ticker(english_name)
            if ticker:
                print(f"✅ [해외] {stock_name} → 티커: {ticker}")
                return {"type": "US", "code": ticker, "name": stock_name}
        
        print(f"❌ '{stock_name}'을 찾을 수 없어요.")
        print(f"💡 힌트: 영문명이나 티커로 직접 입력해보세요!")
        return None
    
    else:
        ticker = get_yahoo_ticker(stock_name)
        if ticker:
            print(f"✅ [해외] {stock_name} → 티커: {ticker}")
            return {"type": "US", "code": ticker, "name": stock_name}
        print(f"❌ '{stock_name}'을 찾을 수 없어요.")
        return None
```

```python
# 뉴스 크롤링 함수
def get_article_body(url):
    headers = {
        "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36",
        "Referer": "https://finance.naver.com"
    }
    try:
        # 1단계: 원래 URL 요청
        response = requests.get(url, headers=headers)
        response.encoding = "euc-kr"
        
        # 2단계: JS 리다이렉트 URL 추출
        if "top.location.href=" in response.text:
            import re
            match = re.search(r"top\.location\.href='(.+?)'", response.text)
            if match:
                real_url = match.group(1)
                response = requests.get(real_url, headers=headers)
                response.encoding = "utf-8"
        
        soup = BeautifulSoup(response.text, "html.parser")
        
        # n.news.naver.com 셀렉터
        selectors = [
            "div#dic_area",        # 네이버 뉴스 본문
            "div.newsct_article",  # 네이버 뉴스 본문 (구버전)
            "div#news_read",
            "div#contents",
        ]
        for selector in selectors:
            body = soup.select_one(selector)
            if body and body.text.strip():
                return body.text.strip()
        return ""
    except:
        return ""

def get_naver_finance_news(stock_info, num_articles=20):
    code = stock_info["code"]
    stock_name = stock_info["name"] 
    headers = {
        "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36",
        "Referer": "https://finance.naver.com"
    }
    results = []
    existing_links = set()
    page = 1
    
    while len(results) < num_articles:
        url = f"https://finance.naver.com/item/news_news.naver?code={code}&page={page}"
        response = requests.get(url, headers=headers)
        response.encoding = "euc-kr"
        soup = BeautifulSoup(response.text, "html.parser")
        articles = soup.select("td.title a")
        
        if not articles:
            break
        
        for article in articles:
            if len(results) >= num_articles:
                break
            title = article.text.strip()
            
            if stock_name not in title: 
                continue
            
            link = "https://finance.naver.com" + article["href"]
            
            if link in existing_links:
                continue
            
            body = get_article_body(link)
            results.append({"title": title, "link": link, "body": body})
            existing_links.add(link)
            time.sleep(0.3)
            print(f"✅ 수집완료 ({len(results)}/{num_articles}): {title[:30]}...")
        
        page += 1
        time.sleep(1)
    
    return results

def get_yahoo_news(ticker, num_articles=20):
    """yfinance로 해외 종목 뉴스 수집"""
    try:
        stock = yf.Ticker(ticker)
        news_data = stock.news
        
        if not news_data:
            print(f"❌ {ticker} 뉴스를 찾을 수 없어요!")
            return []
        
        results = []
        for item in news_data[:num_articles]:
            title = item.get("content", {}).get("title", "")
            link = item.get("content", {}).get("canonicalUrl", {}).get("url", "")
            summary = item.get("content", {}).get("summary", "")
            
            if title:
                results.append({
                    "title": title,
                    "link": link,
                    "body": summary
                })
                print(f"✅ 수집완료: {title[:30]}...")
        
        return results
    except Exception as e:
        print(f"❌ 오류 발생: {e}")
        return []


def get_news(stock_info, num_articles=20):
    if stock_info["type"] == "KR":
        print(f"📰 네이버 금융에서 수집 중...")
        return get_naver_finance_news(stock_info, num_articles)
    else:
        print(f"📰 야후 파이낸스에서 수집 중...")
        return get_yahoo_news(stock_info["code"], num_articles)

def get_stock_chart(stock_info, period="3mo"):
    try:
        if stock_info["type"] == "KR":
            ticker = stock_info["code"] + ".KS"
        else:
            ticker = stock_info["code"]
        
        stock = yf.Ticker(ticker)
        df = stock.history(period=period)
        
        if df.empty:
            return None
        
        fig = go.Figure()
        fig.add_trace(go.Candlestick(
            x=df.index,
            open=df["Open"],
            high=df["High"],
            low=df["Low"],
            close=df["Close"],
            name="주가"
        ))
        
        fig.update_layout(
            title=f"{stock_info['name']} 주가",
            xaxis_rangeslider_visible=False,
            height=400
        )
        
        return fig
    except:
        return None
```

```python
# ChromaDB 벡터 저장
import re
import warnings
warnings.filterwarnings("ignore")

from langchain_huggingface import HuggingFaceEmbeddings
from langchain_community.vectorstores import Chroma
from langchain_core.documents import Document

def store_to_chromadb(news_list, stock_name, stock_code):
    
    splitter = RecursiveCharacterTextSplitter(
        chunk_size=300,
        chunk_overlap=50
    )
    
    docs = []
    for news in news_list:
        if news["body"]:
            chunks = splitter.split_text(news["body"])
            for chunk in chunks:
                docs.append(Document(
                    page_content=chunk,
                    metadata={"title": news["title"], "link": news["link"]}
                ))
    
    if not docs:
        return None
    
    print(f"청킹 완료: 뉴스 {len(news_list)}개 → 청크 {len(docs)}개")
    
    embedding_model = HuggingFaceEmbeddings(
        model_name="BAAI/bge-m3",
        model_kwargs={"device": "cpu"},
        encode_kwargs={"normalize_embeddings": True}
    )
    
    vectorstore = Chroma.from_documents(
        documents=docs,
        embedding=embedding_model,
        persist_directory=f"./chroma_db/{stock_code}",
        collection_name=f"news_{stock_code}"
    )
    
    print(f"✅ {len(docs)}개 청크 벡터 저장 완료!")
    return vectorstore
```

```python
stock_name = input("🔍 종목명을 입력하세요 (예: 삼성전자 / 엔비디아 / TSLA / 005930): ").strip()
stock_info = search_stock(stock_name)

if stock_info:
    news = get_news(stock_info, num_articles=20)
    print(f"\n총 {len(news)}개 기사 수집 완료!")
    
    for i, n in enumerate(news):
        print(f"\n[{i+1}] {n['title']}")
        print(f"     본문길이: {len(n['body'])}자")
    
    vectorstore = store_to_chromadb(news, stock_info["name"], stock_info["code"])
```

```python
#RAG 체인 구성
from langchain_ollama import OllamaLLM
from langchain_core.prompts import PromptTemplate
from langchain_classic.chains import RetrievalQA 

def build_rag_chain(vectorstore, stock_name):
    llm = OllamaLLM(model="llama3.2", temperature=0.1)
    prompt_template = f"""
당신은 주식 뉴스 분석 전문가입니다.
아래 뉴스 기사들 중 {stock_name}과 직접 관련된 내용만 바탕으로 답해주세요.
관련 없는 내용은 무시하고, 반드시 한국어로 핵심만 정리해주세요.

[참고 뉴스]
{{context}}

[질문]
{{question}}

[분석 결과]
"""
    PROMPT = PromptTemplate(
        template=prompt_template,
        input_variables=["context", "question"]
    )
    qa_chain = RetrievalQA.from_chain_type(
        llm=llm,
        chain_type="stuff",
        retriever=vectorstore.as_retriever(search_kwargs={"k": 5}),
        chain_type_kwargs={"prompt": PROMPT},
        return_source_documents=True
    )
    return qa_chain
```

```python
# RAG 체인 생성
chain = build_rag_chain(vectorstore, stock_info["name"])

# 질문
query = input("💬 질문을 입력하세요: ").strip()
result = chain.invoke({"query": query})
print("\n📊 분석 결과:")
print(result["result"])
```
